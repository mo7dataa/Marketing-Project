```python
import pandas as pd
x=pd.read_csv("nykaa_campaign_data.csv")
x.head()
```

```python
x.info()
```

```python
import numpy as np
x["Ci"] = ((x["Clicks"] / x["Impressions"])*100).replace(0, np.nan)
x["Lc"] = ((x["Leads"] / x["Clicks"])*100).replace(0, np.nan)
x["Cl"] = ((x["Conversions"] / x["Leads"])*100).replace(0, np.nan)
x["Cc"] = ((x["Conversions"] / x["Clicks"])*100).replace(0, np.nan)
x.info()
```

```python
x['ROI_check'] = (x['Revenue'] - x['Acquisition_Cost']) / x['Acquisition_Cost']
x[["ROI","ROI_check"]] #ROI SC
x.drop(columns="ROI",inplace=True)
```

```python
x["Ci"].describe()
x["Lc"].describe()
x["Cl"].describe()
x["Cc"].describe()
```

```python
x["Date"]=pd.to_datetime(x["Date"])
x=x.sort_values(by="Date")
x['MonthYear'] = x['Date'].dt.strftime('%B %Y')
x.head()
```

```python
x["Campaign_ID"].duplicated().sum()
x["Campaign_Type"].value_counts()
x["Target_Audience"].value_counts()
x["Channel_Used"].value_counts() # info
x["Language"].value_counts()
x["Channel_Used"].value_counts()
```

```python
x["Duration"].describe() # dur 5:30 eq not outliers
import plotly.express as px
#px.box(x,x="Duration")
print("   ")
```

```python
x.loc[
    (x['Impressions'] >= x['Clicks']) &
    (x['Clicks'] >= x['Leads']) &
    (x['Leads'] >= x['Conversions']),
    'Impressions'
].count()
x.loc[(x["Acquisition_Cost"] >=0) & (x["Revenue"]>=0),"Revenue"].count()
#true
```

```python
x[["Ci", "Lc", "Cl", "Cc","ROI_check"]] = x[["Ci", "Lc", "Cl", "Cc","ROI_check"]].round(2)
```
